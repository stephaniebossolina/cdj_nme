```python
# make an array of numbers 0 to 100 (exclusive)
random_list = np.arange(stop=100)
# print the random_list variable
print(random_list)
```

```python
#doing mathematical operations on our list
#want list to start at 1
#addd 1 to each
random_list = random_list + 1
print(random_list)
```

```python
print(np.log(random_list))
```

```python
#boolean indexing can also be performed using a boolean array
#index array allows us to see where all the odds are
print(random_list % 2 == 0)

even_list_2 = random_list[random_list % 2 == 0]
print(even_list_2)
```

```python
even_list_3 = np.array([])

for i in range(1, 101):
    if i % 2 == 0:
        even_list_3 = np.append(arr = even_list_3, values = i)

print(even_list_3)
```

```python
#make an array of numbers 0 to 200 exlclusive
big_list = np.arange(200)
even_list_A = [2*x for x in big_list if x <= 50] #saves many lines of code
even_list_A = np.array(even_list_A)
print(even_list_A)
```
